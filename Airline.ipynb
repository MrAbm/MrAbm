```python
#Show Bar Graphs representing the Departure Time & Arrival Time.
plt.figure(figsize=(16,4))

plt.subplot(1,2,1)
plt.title("Arrival")
plt.bar(df['arrival_time'].value_counts().index,df['arrival_time'].value_counts().values)
plt.subplot(1,2,2)
plt.title("Departure")
plt.bar(df['departure_time'].value_counts().index,df['departure_time'].value_counts().values)
plt.show()
```

```python
#Q.3. Show Bar Graphs representing the Source City & Destination City.





plt.figure(figsize=(16,4))

plt.subplot(1,2,1)
plt.title("Source")
plt.bar(df['source_city'].value_counts().index,df['source_city'].value_counts().values)
plt.subplot(1,2,2)
plt.title("Destination")
plt.bar(df['destination_city'].value_counts().index,df['destination_city'].value_counts().values)
plt.show()
```
